```python
import torch

!pip install ultralytics
from ultralytics import YOLO

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Using device:", device)

model = YOLO("yolov8n.pt").to(device)
```

```python
from google.colab import drive


from PIL import Image
from io import BytesIO
import requests

drive.mount('/content/drive')
```

```python
img_file = "https://pds.joongang.co.kr/news/component/htmlphoto_mmdata/202505/08/3a9aaa79-0950-4fc1-8d05-64d72826e205.jpg"

resp = requests.get(img_file, timeout=10)
resp.raise_for_status()

img = Image.open(BytesIO(resp.content)).convert("RGB")
```

```python
results = model.predict(img, device=device)
```

```python
for r in results:
    for box in r.boxes:
        cls_id = int(box.cls)
        score = float(box.conf)
        x1, y1, x2, y2 = box.xyxy[0].tolist()
        print(f"class={cls_id}, score={score:.3f}, box=({x1:.1f},{y1:.1f})-({x2:.1f},{y2:.1f})")
```

```python
results[0].save("detected.jpg")
results[0].show()
```
